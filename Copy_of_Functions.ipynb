## 25 Easy Python Functions

```python
# Function 1: Add two numbers
def add(a, b):
  return a + b

print(f"5 + 3 = {add(5, 3)}")
```

```python
# Function 2: Subtract two numbers
def subtract(a, b):
  return a - b

print(f"10 - 4 = {subtract(10, 4)}")
```

```python
# Function 3: Multiply two numbers
def multiply(a, b):
  return a * b

print(f"6 * 7 = {multiply(6, 7)}")
```

```python
# Function 4: Divide two numbers
def divide(a, b):
  if b == 0:
    return "Cannot divide by zero"
  return a / b

print(f"15 / 3 = {divide(15, 3)}")
```

```python
# Function 5: Check if a number is even
def is_even(number):
  return number % 2 == 0

print(f"Is 8 even? {is_even(8)}")
print(f"Is 7 even? {is_even(7)}")
```

```python
# Function 6: Print a greeting message
def greet(name):
  return f"Hello, {name}!"

print(greet("Alice"))
```

```python
# Function 7: Calculate the square of a number
def square(number):
  return number * number

print(f"The square of 9 is {square(9)}")
```

```python
# Function 8: Get the length of a string
def string_length(s):
  return len(s)

print(f"Length of 'Python' is {string_length('Python')}")
```

```python
# Function 9: Convert a string to uppercase
def to_uppercase(s):
  return s.upper()

print(f"'hello' in uppercase is {to_uppercase('hello')}")
```

```python
# Function 10: Find the maximum of two numbers
def find_max(a, b):
  return max(a, b)

print(f"Max of 12 and 20 is {find_max(12, 20)}")
```

```python
# Function 11: Find the minimum of two numbers
def find_min(a, b):
  return min(a, b)

print(f"Min of 12 and 20 is {find_min(12, 20)}")
```

```python
# Function 12: Concatenate two strings
def concatenate_strings(s1, s2):
  return s1 + s2

print(f"'Hello' + 'World' = {concatenate_strings('Hello', 'World')}")
```

```python
# Function 13: Check if an element is in a list
def is_in_list(element, my_list):
  return element in my_list

my_list = [1, 2, 3, 4, 5]
print(f"Is 3 in {my_list}? {is_in_list(3, my_list)}")
print(f"Is 6 in {my_list}? {is_in_list(6, my_list)}")
```

```python
# Function 14: Get the first element of a list
def first_element(my_list):
  if not my_list:
    return None
  return my_list[0]

print(f"First element of [10, 20, 30] is {first_element([10, 20, 30])}")
```

```python
# Function 15: Get the last element of a list
def last_element(my_list):
  if not my_list:
    return None
  return my_list[-1]

print(f"Last element of [10, 20, 30] is {last_element([10, 20, 30])}")
```

```python
# Function 16: Reverse a string
def reverse_string(s):
  return s[::-1]

print(f"'Python' reversed is {reverse_string('Python')}")
```

```python
# Function 17: Calculate the average of a list of numbers
def calculate_average(numbers):
  if not numbers:
    return 0
  return sum(numbers) / len(numbers)

print(f"Average of [1, 2, 3, 4, 5] is {calculate_average([1, 2, 3, 4, 5])}")
```

```python
# Function 18: Check if a string is empty
def is_empty_string(s):
  return len(s) == 0

print(f"Is '' empty? {is_empty_string('')}")
print(f"Is 'abc' empty? {is_empty_string('abc')}")
```

```python
# Function 19: Get the absolute value of a number
def absolute_value(number):
  return abs(number)

print(f"Absolute value of -7 is {absolute_value(-7)}")
```

```python
# Function 20: Repeat a string N times
def repeat_string(s, n):
  return s * n

print(f"'Hi' repeated 3 times: {repeat_string('Hi', 3)}")
```

```python
# Function 21: Check if a number is positive
def is_positive(number):
  return number > 0

print(f"Is 5 positive? {is_positive(5)}")
print(f"Is -2 positive? {is_positive(-2)}")
```

```python
# Function 22: Convert Fahrenheit to Celsius
def fahrenheit_to_celsius(fahrenheit):
  return (fahrenheit - 32) * 5/9

print(f"100 F is {fahrenheit_to_celsius(100):.2f} C")
```

```python
# Function 23: Convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
  return (celsius * 9/5) + 32

print(f"37 C is {celsius_to_fahrenheit(37):.2f} F")
```

```python
# Function 24: Get the sum of a list of numbers
def list_sum(numbers):
  return sum(numbers)

print(f"Sum of [1, 2, 3, 4] is {list_sum([1, 2, 3, 4])}")
```

```python
# Function 25: Get a sublist using slicing
def get_sublist(my_list, start, end):
  return my_list[start:end]

my_list = ['a', 'b', 'c', 'd', 'e']
print(f"Sublist of {my_list} from index 1 to 4 is {get_sublist(my_list, 1, 4)}")
```
